## 1. Python Data Types - Dictionary Object
---
The **Python Dictionary** object is another collection object. Like lists, dictionaries are mutable, so values can be added and removed.

Unlike a List, which is indexed with the offset integer, a Dictionary is indexed with a unique key. Dictionaries can be thought of as a set of ***key: value pairs***.

Python Lists are *ordered* by the offset value, meaning they can be sorted and reordered. Python Dictionaries are officially classified as unordered, but since Python 3.7, the ***insertion order is maintained***. You can use the sorted() function to sort a dictionary for output, but there is no .sort() method that changes the dictionary itself. 

Dictionary **keys** can be any immutable data type - e.g. strings, integers, floats. Dictionary **values** can be any type of object. Each value in the dictionary is accessed using its unique key. Dictionaries in Python are similar to *hash tables/maps* in other languages, and allow direct access to the *value* associated with the *key*, without having to iterate through a collection. When using large collections, hash tables can give significant speed advantages over indexed collections.

To associate an object to a key the following format is used:
<br>

 **\<key>:\<value>** A dictionary is composed of key/value pairs such as the following:

## 2. Iteration of Dictionaries
---
If you need to perform tasks on each item of a Dictionary, you can use a loop, using the built in Python **for .. in** construct (similar to lists and strings).

By default the loop iterates over the keys of the Dictionary, and you have to access the value using the key. Try the following: 

````python
print(dic)
for key in dic:
    print(f"{key} -> {dic[key]}")
````

This is the same as iterating over the list of keys returned by the **.keys()** method. Try:  

````python
for key in dic.keys():
    print(f"{key} -> {dic[key]}")
````

The previous method **is not the most efficient** to access the values of a dictionary as the dictionary is read twice for each iteration:
1. to get the key
2. to get the value for the key

<span style = 'color:red'>***It is better to use the .items() method***</span> which needs to access the dictionary only once for each iteration, returning a (key, value) tuple. We can then assign the tuple contents to two variables and use them. Try:

~~~python
for key, value in dic.items():
    print(f"{key} -> {value}")
~~~

````python
for key, value in dic.items():
    print(f"{key} -> {value}")
````

## 12. Hash signatures - Python Cyber Security Applications
---
Hash signatures can be used for authentication of users and to ensure the integrity of messages, files, and even entire digital media.

For example, PyPi, the Python Package Index, includes hashes for every file that can be downloaded. After downloading a file, you should check that the hash of your copy of the file matches the one given on the website. If this is not the case, you know that your copy of the file has been compromised, for example, through a man-in-the-middle attack.

To generate a hash signature, a hash algorithm is used. In the example above, you can see SHA256 and a second hash algorithm, BLAKE2b-256.

**MD5** is still one of the most common hash algorithms used, though it is no longer shown in PyPi.  
In Python this is implemented as the **md5()** function in the **hashlib** standard library module.  
To get a printable hex encoding of the generated hash, you then need to apply the **.hexdigest()** method.

Hash functions *cannot be called directly on strings*.  
The *string needs to be encoded to bytes* using the **.encode()** string method before you can hash it.  
utf-8 is the most widely used encoding worldwide and is the default in Python.  

Try out the code below:

````python
import hashlib
md5hash = hashlib.md5("Petra".encode("utf-8"))
print(md5hash.hexdigest())
````

Because utf-8 is the default, you could have written the second line as:
~~~python
md5hash = hashlib.md5("Petra".encode())
~~~
